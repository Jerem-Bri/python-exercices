```python
# Chargement du jeu de données
import seaborn as sns
import pandas as pd

titanic = sns.load_dataset("titanic")
```

```python
# 1. Nombre de passagers de moins de 18 ans
mineurs = (titanic["age"] < 18).sum()
print("Moins de 18 ans :", mineurs)  # 113
```

```python
# 2. Les 5 billets les plus chers, avec la classe et le port d'embarquement
max_billets = (
    titanic.sort_values("fare", ascending=False)
    .head(5)[["fare", "class", "embark_town"]]
)

print(max_billets)
```

```python
# 3. Part des femmes parmi les passagers de 1re classe
passagers_1re = titanic[titanic["class"] == "First"]
part_femmes_1re = (passagers_1re["sex"] == "female").mean()
print(f"Part de femmes en 1re classe : {part_femmes_1re:.1%}")
```
