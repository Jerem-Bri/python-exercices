```python
# Chargement du jeu de données
import seaborn as sns
import pandas as pd

titanic = sns.load_dataset("titanic")
```

```python
# 1. Nombre de lignes et de colonnes
print(titanic.shape)
```

```python
# 2. Type de chaque colonne
print(titanic.dtypes)
```

```python
# 3. Nombre de valeurs manquantes par colonne
print(titanic.isna().sum())
```

```python
# 4. Répartition des passagers par classe et par port d'embarquement
# (les valeurs manquantes de embark_town sont affichées dans une colonne NaN)
pd.crosstab(titanic["class"], titanic["embark_town"], dropna=False, margins=True)
```
